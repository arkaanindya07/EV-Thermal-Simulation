```python
# EV Battery Thermal Simulation: 60-Second Hard Acceleration

# 1. Physical Variables (The Setup)
current_temp = 25.0    # Starting temperature in Celsius
current = 150.0        # Current draw during acceleration in Amps
resistance = 0.005     # Internal electrical resistance in Ohms
mass = 1.2             # Mass of the battery cell in kg
specific_heat = 850    # Specific heat capacity in J/(kg*C)
cooling_rate = 5.0     # Heat removed by the cooling system per second

print("Starting 60-second acceleration simulation...")
print("---------------------------------------------")

# 2. The Simulation Loop
for second in range(1, 61):

    # Calculate heat generated by electrical resistance (I^2 * R)
    heat_generated = (current ** 2) * resistance

    # Calculate net heat (Heat Generated - Heat Removed)
    net_heat = heat_generated - cooling_rate

    # Calculate temperature change for this second
    temp_change = net_heat / (mass * specific_heat)

    # Update the battery's current temperature
    current_temp = current_temp + temp_change

    # Print the temperature to the screen every 10 seconds
    if second % 10 == 0:
        print(f"Time: {second}s | Battery Temp: {current_temp:.2f} °C")

print("---------------------------------------------")

# 3. Safety Check
if current_temp > 60.0:
    print("WARNING: Thermal limit exceeded. Cooling system upgrade required.")
else:
    print("System safe. Battery remained under 60°C.")
```
